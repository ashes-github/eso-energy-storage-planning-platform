# Phase 1 results
Run the pipeline first, then inspect its saved outputs here. This notebook performs reporting only. Candidate labels are uncalibrated demand-pattern screening hypotheses.


```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'configs/phase1.json').exists())
run = root / 'output/phase1_signed_summer_weekdays'
print(json.loads((run / 'run_status.json').read_text()))
display(pd.read_csv(run / 'comparison.csv').sort_values('heldout_gsp_withheld_period_mse_standardized'))
```

```python
audit = json.loads((run / 'data_audit.json').read_text())
display(audit)
screening = pd.read_csv(run / 'screening.csv')
display(screening.groupby('screening_status').size())
display(pd.read_csv(run / 'excluded_gsps.csv').head())
display(Image(filename=str(run / 'screening_map.png')))
```

```python
display(Image(filename=str(run / 'pca/latent.png')))
models = sorted(run.glob('gplvm*/latent.png'))
if models:
    display(Image(filename=str(models[0])))
```
